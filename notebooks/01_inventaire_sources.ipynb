```python
import sys
from pathlib import Path
ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from src.data.paths import ProjectPaths, get_project_root, CSV_FILES
from src.data.registre import build_source_column_index
from src.data.load import load_csv

paths = ProjectPaths(root=get_project_root())
index = build_source_column_index(paths)
index
```

```python
rows = []
for name in CSV_FILES:
    df = load_csv(paths, name, from_raw=False)
    rows.append({
        "fichier": name,
        "n_lignes": len(df),
        "n_colonnes": df.shape[1],
        "colonnes": ", ".join(df.columns),
        "null_rate_max": float(df.isna().mean().max()) if len(df) else 0.0,
    })
import pandas as pd
pd.DataFrame(rows)
```

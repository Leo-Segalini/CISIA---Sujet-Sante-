```python
import matplotlib.pyplot as plt
usage_cols = ["usage_score_sortie", "usage_score_tele"]
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, col in zip(axes, usage_cols):
    if col in reg.columns:
        vc = reg[col].value_counts()
        ax.bar(vc.index, vc.values, color=["#2d6765", "#ba8c2e", "#ba1a1a"][: len(vc)])
        ax.set_title(col.replace("_", " "))
        ax.set_ylabel("Nb colonnes")
plt.tight_layout()
plt.show()
```

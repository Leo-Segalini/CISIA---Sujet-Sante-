# 03 — Registre colonnes & sensibilité RGPD

Le registre (`docs/registres/registre_colonnes.csv`) documente chaque colonne source :
- **sensibilité** (identité, santé art.9, proxy socio…)
- **usage_score_sortie** / **usage_score_tele** (`autorise`, `flag_only`, `exclu`)
- **justification** métier


```python
import runpy
from pathlib import Path
_sp = next(a / "setup_path.py" for a in [Path.cwd(), *Path.cwd().parents] if (a / "setup_path.py").is_file())
runpy.run_path(str(_sp))

%matplotlib inline
from _utils.bootstrap import paths
import pandas as pd
from IPython.display import display
from src.data.registre import (
    assert_justifications,
    assert_registre_covers_sources,
    load_registre,
)

reg = load_registre(paths)
print(f"Colonnes registre : {len(reg)}")
assert_registre_covers_sources(paths)
assert_justifications(reg)
print("✅ Registre couvre toutes les sources et justifications OK")
```

```python
# Répartition sensibilité
if "sensibilite" in reg.columns:
    display(reg["sensibilite"].value_counts().rename("n").reset_index())
```

```python
# Colonnes exclues du score sortie
exclu = reg.loc[
    reg["usage_score_sortie"] == "exclu",
    ["fichier_source", "colonne", "sensibilite", "justification_hopital"],
]
print(f"Exclues score sortie : {len(exclu)}")
display(exclu.head(20))
```

```python
import matplotlib.pyplot as plt
from IPython.display import display

usage_cols = ["usage_score_sortie", "usage_score_tele"]
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, col in zip(axes, usage_cols):
    if col in reg.columns:
        vc = reg[col].value_counts()
        ax.bar(vc.index.astype(str), vc.values, color=["#2d6765", "#ba8c2e", "#ba1a1a"][: len(vc)])
        ax.set_title(col.replace("_", " "))
        ax.set_ylabel("Nb colonnes")
fig.tight_layout()
display(fig)
plt.close(fig)
```
